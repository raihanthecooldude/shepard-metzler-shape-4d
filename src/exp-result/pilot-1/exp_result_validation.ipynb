```python
import pandas as pd
import numpy as np
from math import comb
```

```python
df = pd.read_csv('data_with_feedback.csv')

diff = np.abs(df['deg1'] - df['deg2'])
deg_diff = np.minimum(diff, 360 - diff)
df['angle_diff'] = deg_diff

df.head()
```

### all angles + random shapes

```python
vec_random_all = df['is_correct'].astype(int).tolist()

print(vec_random_all)
print("Length:", len(vec_random_all))
```

```python
number_correct_random_all = sum(vec_random_all)
number_incorrect_random_all = 100 - number_correct_random_all

print("Correct:", number_correct_random_all)
print("Incorrect:", number_incorrect_random_all)
```

```python
p_value_random_all = sum(comb(100, k) * 0.5**100 for k in range(number_correct_random_all, 101))
print("p value:", p_value_random_all)
```

### filtered angle + random shapes

```python
filtered_angle_value = 45
```

```python
angle_filtered_df = df[deg_diff > filtered_angle_value].copy()

angle_filtered_df.head()
```

```python
vec_angle_filtered = angle_filtered_df['is_correct'].astype(int).tolist()

print(vec_angle_filtered)
print("Length:", len(vec_angle_filtered))
```

```python
number_correct_random_angle = sum(vec_angle_filtered)
number_incorrect_random_angle = len(vec_angle_filtered) - number_correct_random_angle

print("Correct:", number_correct_random_angle)
print("Incorrect:", number_incorrect_random_angle)
```

```python
p_value_random_angle = sum(comb(len(vec_angle_filtered), k) * (0.5**len(vec_angle_filtered)) for k in range(number_correct_random_angle, len(vec_angle_filtered)+1))
print("p value:", p_value_random_angle)
```
